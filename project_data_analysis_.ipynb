```python
import pandas as pd


# ============================================================
# ΒΗΜΑ 1 & 2: Η κλάση Employee με attributes + display_info
# ============================================================
class Employee:
    def __init__(self, first_name, last_name, age, salary, department):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.salary = salary
        self.department = department

    def display_info(self):
        return f"{self.first_name} {self.last_name}, Ηλικία: {self.age}, Μισθός: €{self.salary:.2f}, Τμήμα: {self.department}"

    # ============================================================
    # ΒΗΜΑ 10: Επέκταση της κλάσης με give_bonus, change_department
    # ============================================================
    def give_bonus(self, bonus_amount):
        self.salary = self.salary + bonus_amount

    def change_department(self, new_department):
        self.department = new_department


# ============================================================
# ΒΗΜΑ 3: Ανάγνωση CSV με pandas
# ============================================================
df = pd.read_csv("pythondata.csv")


# ============================================================
# ΒΗΜΑ 4: Προεπισκόπηση δεδομένων
# ============================================================
print(df.head())
print(df.columns)


# ============================================================
# ΒΗΜΑ 5: Καθαρισμός δεδομένων (NaN -> mean, age -> int)
# ============================================================
print(df.isnull().sum())   # πριν τον καθαρισμό

mean_age = df['age'].mean()
mean_salary = df['salary'].mean()

df['age'] = df['age'].fillna(mean_age)
df['salary'] = df['salary'].fillna(mean_salary)

df['age'] = df['age'].astype(int)

print(df.isnull().sum())   # μετά τον καθαρισμό, πρέπει να είναι όλα 0


# ============================================================
# ΒΗΜΑ 6: Στατιστικά στοιχεία με describe()
# ============================================================
print(df[['age', 'salary']].describe())


# ============================================================
# ΒΗΜΑ 7: Δημιουργία Employee objects από το DataFrame
# ============================================================
employees = []

for index, row in df.iterrows():
    emp = Employee(row['first_name'], row['last_name'], row['age'], row['salary'], row['department'])
    employees.append(emp)

print(len(employees))
print(employees[0].display_info())


# ============================================================
# ΒΗΜΑ 8: Εμφάνιση όλων των υπαλλήλων με loop
# ============================================================
for emp in employees:
    print(emp.display_info())


# ============================================================
# ΒΗΜΑ 9: Φιλτράρισμα υπαλλήλων άνω των 30 ετών
# ============================================================
older_than_30 = df[df['age'] > 30]
print(older_than_30)
print(len(older_than_30))


# ============================================================
# ΒΗΜΑ 10 (δοκιμή): give_bonus & change_department σε δράση
# ============================================================
test_emp = employees[0]
print(test_emp.display_info())

test_emp.give_bonus(500)
print(test_emp.display_info())

test_emp.change_department("Management")
print(test_emp.display_info())

import psycopg2

conn = psycopg2.connect(
    host="ep-calm-cell-b1p273zj-pooler.c-5.eu-central-1.aws.neon.tech",
    port=5432,
    dbname="neondb",
    user="neondb_owner",
    password="npg_9uqIBb7CNAKi",
    sslmode="require"
)

cursor = conn.cursor()

print("Συνδέθηκα επιτυχώς!")

create_table_query = """
CREATE TABLE IF NOT EXISTS employees (
    id SERIAL PRIMARY KEY,
    first_name VARCHAR(100),
    last_name VARCHAR(100),
    age INTEGER,
    salary FLOAT,
    department VARCHAR(100)
);
"""

cursor.execute(create_table_query)
conn.commit()

print("Ο πίνακας δημιουργήθηκε επιτυχώς!")

cursor.execute("DELETE FROM employees;")
conn.commit()

print("Ο πίνακας αδειάστηκε!")

insert_query = """
INSERT INTO employees (first_name, last_name, age, salary, department)
VALUES (%s, %s, %s, %s, %s);
"""

for index, row in df.iterrows():
    cursor.execute(insert_query, (row['first_name'], row['last_name'], row['age'], row['salary'], row['department']))

conn.commit()

print("Όλα τα δεδομένα εισήχθησαν επιτυχώς!")

cursor.execute("SELECT COUNT(*) FROM employees;")
print(cursor.fetchone())

```
